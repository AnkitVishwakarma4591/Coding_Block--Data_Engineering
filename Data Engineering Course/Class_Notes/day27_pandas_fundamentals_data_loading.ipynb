## 1. Import Libraries


````python
import pandas as pd
import numpy as np

print("pandas version:", pd.__version__)
````

## 2. Series: 1D Labeled Array

A Series is a one-dimensional labeled array.

It can be used to represent one column of data.


````python
sales_series = pd.Series([1000, 1500, 1200, 1800], name="sales")

print(sales_series)
print(type(sales_series))
````

## 3. Series with Custom Index


````python
daily_sales = pd.Series(
    data=[1000, 1500, 1200, 1800],
    index=["Monday", "Tuesday", "Wednesday", "Thursday"],
    name="daily_sales"
)

print(daily_sales)
print(daily_sales["Tuesday"])
````

## 4. DataFrame: 2D Tabular Structure

A DataFrame is a two-dimensional table with rows and columns.

It is the most commonly used Pandas object in data analysis and data engineering.


````python
orders_df = pd.DataFrame({
    "order_id": [1001, 1002, 1003, 1004],
    "customer_id": [501, 502, 503, 501],
    "amount": [250.50, -100.00, 500.00, 1200.00],
    "status": ["completed", "completed", "pending", "completed"]
})

print(orders_df)
print(type(orders_df))
````

## 5. Selecting a Column Returns a Series


````python
amount_series = orders_df["amount"]

print(amount_series)
print(type(amount_series))
````

## 6. Selecting Multiple Columns Returns a DataFrame


````python
selected_columns = orders_df[["order_id", "amount"]]

print(selected_columns)
print(type(selected_columns))
````

# Creating Sample Files for Data Loading


## 7. Create Sample CSV File


````python
orders_data = pd.DataFrame({
    "order_id": [1001, 1002, 1003, 1004, 1005, 1006],
    "customer_id": [501, 502, 503, 501, 504, 505],
    "amount": [250.50, -100.00, 500.00, 1200.00, 0.00, 750.00],
    "status": ["completed", "completed", "pending", "completed", "completed", "completed"],
    "city": ["Delhi", "Mumbai", "Bangalore", "Delhi", "Pune", "Chennai"]
})

orders_data.to_csv("orders.csv", index=False)

print("orders.csv created")
````

## 8. Create Sample JSON File


````python
products_data = pd.DataFrame({
    "product_id": [201, 202, 203, 204],
    "product_name": ["Laptop", "Phone", "Shoes", "Keyboard"],
    "category": ["Electronics", "Electronics", "Fashion", "Electronics"],
    "price": [80000, 40000, 3000, 1500],
    "stock": [15, 5, 50, 8]
})

products_data.to_json("products.json", orient="records", indent=2)

print("products.json created")
````

iloc selects rows and columns by integer position.

## 25. loc: Label Based Selection

loc selects rows and columns by labels.


````python
print(orders_csv.loc[0])
print(orders_csv.loc[0:3, ["order_id", "amount", "status"]])
````

## 26. Difference Between loc and iloc


````python
print("Using iloc:")
print(orders_csv.iloc[0:3, 0:2])

print("Using loc:")
print(orders_csv.loc[0:3, ["order_id", "customer_id"]])
````

# Boolean Indexing


## 30. Filter Rows Where Amount Is Greater Than 1000


````python
high_value_orders = orders_csv[orders_csv["amount"] > 1000]

print(high_value_orders)
````

## 31. Filter Completed Orders


````python
completed_orders = orders_csv[orders_csv["status"] == "completed"]

print(completed_orders)
````

## 32. Filter with Multiple Conditions

Use:
- & for AND
- | for OR
- parentheses around each condition


````python
valid_completed_orders = orders_csv[
    (orders_csv["amount"] > 0) &
    (orders_csv["status"] == "completed")
]

print(valid_completed_orders)
````

## 33. OR Condition


````python
delhi_or_mumbai_orders = orders_csv[
    (orders_csv["city"] == "Delhi") |
    (orders_csv["city"] == "Mumbai")
]

print(delhi_or_mumbai_orders)
````

## 34. Filter Using isin()


````python
selected_cities = ["Delhi", "Chennai"]

city_orders = orders_csv[orders_csv["city"].isin(selected_cities)]

print(city_orders)
````

## 35. Data Engineering Use Case: Clean Orders Dataset

Rules:
- amount > 0
- status == completed
- customer_id is not missing


````python
clean_orders = orders_csv[
    (orders_csv["amount"] > 0) &
    (orders_csv["status"] == "completed") &
    (orders_csv["customer_id"].notna())
]

print(clean_orders)
````

## 36. Data Engineering Use Case: High Value Customer Orders


````python
high_value_clean_orders = clean_orders[clean_orders["amount"] >= 700]

print(high_value_clean_orders)
````

# Parquet Context for Data Engineering


## 43. CSV vs Parquet Quick Comparison

CSV:
- row-based text format
- larger file size
- no built-in schema preservation
- slower for analytical column reads

Parquet:
- columnar format
- compressed
- stores schema
- efficient for analytics
- widely used with Spark, Hive, Databricks, BigQuery, Athena, and lakehouse systems


````python
print("CSV file: orders.csv")
print("Parquet file: orders.parquet or valid_orders.parquet")
````

## 44. Compare File Sizes


````python
from pathlib import Path

csv_size = Path("orders.csv").stat().st_size

print("orders.csv size in bytes:", csv_size)

if Path("orders.parquet").exists():
    parquet_size = Path("orders.parquet").stat().st_size
    print("orders.parquet size in bytes:", parquet_size)
else:
    print("orders.parquet not available")
````

# Interview Questions


## 51. What is the difference between a Series and a DataFrame?

A Series is a one-dimensional labeled array.

A DataFrame is a two-dimensional tabular structure made of rows and columns.

A single column from a DataFrame is usually a Series.

Multiple columns from a DataFrame form another DataFrame.


## 52. How do you filter a DataFrame to show only rows where Sales > 1000?

Example:

```python
filtered_df = df[df["Sales"] > 1000]
```

For multiple conditions:

```python
filtered_df = df[(df["Sales"] > 1000) & (df["Status"] == "completed")]
```


## 53. Why is Parquet preferred over CSV in Data Engineering pipelines?

Parquet is preferred because:
- it is columnar
- it supports compression
- it stores schema information
- it is faster for analytical queries
- it allows query engines to read only required columns
- it works well with big data tools like Spark, Hive, Databricks, Athena, and lakehouse systems

CSV is simple and readable, but inefficient for large-scale analytics.


## 54. What is the difference between loc and iloc?

loc:
- label-based selection
- uses row labels and column names

iloc:
- integer-position based selection
- uses numeric row and column positions
