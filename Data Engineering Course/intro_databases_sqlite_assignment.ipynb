```python
import requests
import pandas as pd
import sqlite3
from IPython.display import display
```

```python
# Fetching API's
users = requests.get("https://fakestoreapi.com/users/").json()
products = requests.get("https://fakestoreapi.com/products/").json()
carts = requests.get("https://fakestoreapi.com/carts").json()
```

## convert **users** row data into structured dataframe

```python
# Convert to dataframe
df = pd.DataFrame(users)

# Drop irrelevant columns
df.columns

df.drop(columns=['address','__v'], inplace=True)
df.head()
```

```python
df['firstname'] = df['name'].apply(lambda x: x['firstname'])
df['lastname'] = df['name'].apply(lambda x: x['lastname'])

# Drop the original 'name' column
df.drop(columns=['name'], inplace=True)

# Assign the processed DataFrame to users_data
users_data = df.copy()

# Display the first few rows of the updated users_data DataFrame
print(users_data.head())
users_data.to_csv('users_data.csv', index=False)
```

## convert **product** row data into structured dataframe

```python
# Convert to dataframe
df = pd.DataFrame(products)
df.columns
```

```python
df.head()
```

```python
# Drop irrelevant columns
df.drop(columns=['description', 'image'], inplace=True)
df.head()
```

```python
df['rate'] = df['rating'].apply(lambda x: x['rate'])
df['count'] = df['rating'].apply(lambda x: x['count'])

# Drop the original 'rating' column
df.drop(columns=['rating'], inplace=True)
```

```python
df.head()
```

```python
df.to_csv('products_data.csv', index=False)
```

## convert **carts** row data into structured dataframe

```python
df = pd.DataFrame(carts)
df.columns
```

```python
# Drop irrelevant columns
df.drop(columns='__v', inplace=True)
df.head()
```

```python
# convert date column datatype from object to datetime
df['date'] = df['date'].apply(lambda x: x.split('T')[0])
df['date'] = pd.to_datetime(df['date'])
df.info()
```

```python
df['product_id'] = df['products'].apply(lambda x: [item['productId'] for item in x])
df['quantity'] = df['products'].apply(lambda x: [item['quantity'] for item in x])

```

```python
# Drop the original 'products' column
df.drop(columns=['products'], inplace=True)
```

```python
df.head()
```

```python
df_carts_flat = df.explode(['product_id', 'quantity'])

df_carts_flat.rename(columns={'id': 'cart_id', 'product_id': 'product_id_in_cart'}, inplace=True)

# Convert product_id and quantity to numeric types
df_carts_flat['product_id_in_cart'] = pd.to_numeric(df_carts_flat['product_id_in_cart'])
df_carts_flat['quantity'] = pd.to_numeric(df_carts_flat['quantity'])

df_carts_flat.to_csv('carts_data.csv', index=False)

display(df_carts_flat.head())
```

------------------------------------------------------------------
 ### **Loading All the Cleaning Dataset**
-----------------------------------------------------------------

```python
users = pd.read_csv('users_data.csv')
products = pd.read_csv('products_data.csv')
carts = pd.read_csv('carts_data.csv')
```

------------------------------------------------------------------
 ### **Create an In-Memory SQLite Database**
-----------------------------------------------------------------

```python
conn = sqlite3.connect(":memory:")

print("SQLite in-memory database created")
```

------------------------------------------------------------------
 ### **Load Pandas DataFrames into SQLite Tables**
-----------------------------------------------------------------

```python
users.to_sql("users", conn, if_exists="replace", index=False)
products.to_sql("products", conn, if_exists="replace", index=False)
carts.to_sql("carts", conn, if_exists="replace", index=False)

print("Tables loaded into SQLite")
```

------------------------------------------------------------------
 ### **Task 01 : Find all the users who have never placed an order**
-----------------------------------------------------------------

```python
users_no_orders = pd.read_sql_query(
    """
      SELECT
          u.id,
          u.email,
          u.username,
          u.firstname,
          u.lastname
      FROM users u
      LEFT JOIN carts c ON u.id = c.userId
      WHERE c.userId IS NULL
      """, conn)
print("Users who have never placed an order:")
display(users_no_orders)
```

------------------------------------------------------------------
 ### **Task 02 : Find all the products which has no orders**
-----------------------------------------------------------------

```python
query_products_no_orders = """
SELECT
    p.id,
    p.title,
    p.price,
    p.category,
    p.rate,
    p.count
FROM products p
LEFT JOIN carts c ON p.id = c.product_id_in_cart
WHERE c.product_id_in_cart IS NULL
"""

products_no_orders_sql = pd.read_sql_query(query_products_no_orders, conn)
print("Products with no orders:")
display(products_no_orders_sql)
```

------------------------------------------------------------------
### **Task 03: Find top 5 products with orders**
------------------------------------------------------------------

```python
query_top_5_products = """
SELECT
    p.id AS product_id,
    p.title,
    p.category,
    SUM(c.quantity) AS total_quantity_ordered
FROM products p
JOIN carts c ON p.id = c.product_id_in_cart
GROUP BY p.id, p.title, p.category
ORDER BY total_quantity_ordered DESC
LIMIT 5
"""

top_5_products = pd.read_sql_query(query_top_5_products, conn)
print("Top 5 products with orders:")
display(top_5_products)
```

------------------------------------------------------------------
### **Task 04: Join Users with Orders**
------------------------------------------------------------------

```python
query_users_with_orders = """
SELECT
    u.id AS user_id,
    u.firstname,
    u.lastname,
    u.email,
    c.cart_id,
    c.date AS order_date,
    c.product_id_in_cart,
    c.quantity
FROM users u
JOIN carts c ON u.id = c.userId
ORDER BY u.id, c.date
"""

users_with_orders = pd.read_sql_query(query_users_with_orders, conn)
print("Users joined with their orders:")
display(users_with_orders.head(10))
```
