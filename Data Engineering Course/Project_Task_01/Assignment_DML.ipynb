### Create one final SQL report with the following columns. Use aliases to make the output readable.
-------------------------------------------------------------------------------
### **customer_id**
### **customer_name**
### **city**
### **total_completed_orders**
### **total_revenue**
### **average_order_value**
-------------------------------------------------------------------------------

````python
q("""
SELECT
    c.customer_id AS customer_id,
    c.customer_name AS customer_name,
    c.city AS city,
    COUNT(o.order_id) AS total_completed_orders,
    SUM(o.amount) AS total_revenue,
    AVG(o.amount) AS average_order_value
FROM
    customers AS c
JOIN
    orders AS o ON c.customer_id = o.customer_id
WHERE
    o.status = 'completed'
    AND o.amount IS NOT NULL
    AND o.amount > 0
GROUP BY
    c.customer_id, c.customer_name, c.city
ORDER BY
    total_revenue DESC;
""")
````

## Required Interview Answers

### What is the execution order of a SQL query? Explain FROM, WHERE, SELECT, ORDER BY in simple terms.

The logical execution order of a SQL query is generally as follows, though the database optimizer might physically execute it differently:

1.  **FROM**: Identifies the tables involved in the query and how they are joined.
    *   **Simple term**: This is where you tell the database *which tables to look at*.
2.  **WHERE**: Filters rows based on specified conditions.
    *   **Simple term**: This is where you tell the database *which rows to keep* based on certain criteria.
3.  **GROUP BY**: Groups rows that have the same values in specified columns into summary rows.
    *   **Simple term**: This is where you tell the database to *combine similar rows* for aggregation (e.g., sum, count).
4.  **HAVING**: Filters groups based on specified conditions, similar to WHERE but for grouped data.
    *   **Simple term**: This is where you tell the database *which combined groups to keep*.
5.  **SELECT**: Specifies which columns or expressions to return.
    *   **Simple term**: This is where you tell the database *what information to show* from the kept rows/groups.
6.  **ORDER BY**: Sorts the result set based on specified columns.
    *   **Simple term**: This is where you tell the database *how to sort the final results*.

### How do you find all records where the name starts with 'A' and ends with 'Z' or 'z'?

You would use the `LIKE` operator with wildcards. For example:

```sql
SELECT * FROM customers
WHERE customer_name LIKE 'A%Z' OR customer_name LIKE 'A%z';
```

### Why should NULL values be handled using IS NULL instead of = NULL?

`NULL` in SQL represents an unknown or missing value, not a specific data value. Therefore, it cannot be directly compared using standard comparison operators like `=`. The expression `NULL = NULL` does not evaluate to true; instead, it evaluates to `UNKNOWN` (or `NULL` depending on context and SQL dialect). To correctly check for the presence or absence of a `NULL` value, you must use the special operators `IS NULL` or `IS NOT NULL`.

### Why is UPDATE or DELETE without WHERE dangerous in production databases?

Using `UPDATE` or `DELETE` statements without a `WHERE` clause is extremely dangerous in a production database because:

*   **UPDATE without WHERE**: It will modify *every single row* in the specified table. This can lead to mass data corruption, changing all values in a column or set of columns across the entire table, making it unusable and potentially violating data integrity rules.
*   **DELETE without WHERE**: It will delete *every single row* from the specified table. This results in the complete loss of all data in that table, which can be catastrophic for an application or business process that relies on that data. Recovery typically requires restoring from a backup, leading to significant downtime and potential data loss since the last backup.
