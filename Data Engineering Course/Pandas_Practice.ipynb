```python
import requests
import pandas as pd
```

```python
product = requests.get('https://fake-store-api.mock.beeceptor.com/api/products').json()
```

```python
product_id, name, description, price, unit, image, discount, availability, brand, category, rating = [], [], [], [], [], [], [], [], [], [], []
```

```python
for p in product:
  product_id.append(p['product_id'])
  name.append(p['name'])
  description.append(p['description'])
  price.append(p['price'])
  unit.append(p['unit'])
  image.append(p['image'])
  discount.append(p['discount'])
  availability.append(p['availability'])
  brand.append(p['brand'])
  category.append(p['category'])
  rating.append(p['rating'])
```

```python
data = pd.DataFrame({
    'product_id': product_id,
    'name': name,
    'description': description,
    'price': price,
    'unit': unit,
    'image': image,
    'discount': discount,
    'availability': availability,
    'brand': brand,
    'category': category,
    'rating': rating
})
```

```python
print(data)
```

```python
data.to_csv('product.csv', index=False)
data.to_excel('product.xlsx', index=False)
data.to_json('product.json', orient='records', indent=4)
data.to_parquet('product.parquet', index=False)
```
