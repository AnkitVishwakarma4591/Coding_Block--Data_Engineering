```python
# Cell 2: Define dataset URLs
jan_url  = "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2024-01.parquet"
feb_url  = "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2024-02.parquet"
zone_url = "https://d37ci6vzurychx.cloudfront.net/misc/taxi_zone_lookup.csv"
```

```python
# Only load the columns we need (saves memory)
trip_cols = [
    "tpep_pickup_datetime", "tpep_dropoff_datetime",
    "PULocationID", "DOLocationID",
    "passenger_count", "trip_distance",
    "fare_amount", "total_amount", "payment_type"
]
```

```python
print("Loading January 2024 data...")
jan = pd.read_parquet(jan_url, columns=trip_cols)
print(f"January loaded: {jan.shape}")
```

```python
print("Loading February 2024 data...")
feb = pd.read_parquet(feb_url, columns=trip_cols)
print(f"February loaded: {feb.shape}")
```

```python
print("Loading Taxi Zone Lookup...")
zones = pd.read_csv(zone_url)
print(f"Zones loaded: {zones.shape}")
```

```python
trips = pd.concat([jan, feb], ignore_index=True)
print(f"Combined trips shape: {trips.shape}")
print(f"Jan rows: {len(jan):,} | Feb rows: {len(feb):,} | Total: {len(trips):,}")
```

```python
# Ensure datetime types
trips["tpep_pickup_datetime"]  = pd.to_datetime(trips["tpep_pickup_datetime"])
trips["tpep_dropoff_datetime"] = pd.to_datetime(trips["tpep_dropoff_datetime"])
```

```python
# Derived columns
trips["pickup_month"]   = trips["tpep_pickup_datetime"].dt.month          # 1=Jan, 2=Feb
trips["pickup_date"]    = trips["tpep_pickup_datetime"].dt.date           # YYYY-MM-DD
trips["pickup_hour"]    = trips["tpep_pickup_datetime"].dt.hour           # 0-23
trips["trip_minutes"]   = (
    (trips["tpep_dropoff_datetime"] - trips["tpep_pickup_datetime"])
    .dt.total_seconds() / 60
)
```

```python
print("New features created:")
print(trips[["tpep_pickup_datetime", "pickup_month", "pickup_date", "pickup_hour", "trip_minutes"]].head(10))
print(f"\ntrip_minutes stats:\n{trips['trip_minutes'].describe()}")
```

```python
raw_rows = len(trips)
trips_clean = trips[
    (trips["total_amount"]  > 0) &
    (trips["trip_distance"] > 0) &
    (trips["trip_minutes"]  > 0)
].copy()
```

```python
cleaned_rows = len(trips_clean)
removed      = raw_rows - cleaned_rows
```

```python
print(f"Raw rows         : {raw_rows:>10,}")
print(f"Rows removed     : {removed:>10,}  ({removed/raw_rows*100:.2f}%)")
print(f"Remaining rows   : {cleaned_rows:>10,}")
```

```python
pickup_zones = zones.rename(columns={
    "LocationID": "PULocationID",
    "Borough"   : "pickup_borough",
    "Zone"      : "pickup_zone"
})
```

```python
before_join = len(trips_clean)
```

```python
trips_clean = pd.merge(
    trips_clean,
    pickup_zones[["PULocationID", "pickup_borough", "pickup_zone"]],
    on="PULocationID",
    how="left"
)
```

```python
unmatched_pickup = trips_clean["pickup_borough"].isnull().sum()
print(f"Rows before pickup join    : {before_join:,}")
print(f"Unmatched pickup zone rows : {unmatched_pickup:,}")
print(trips_clean[["PULocationID", "pickup_borough", "pickup_zone"]].head())
```

```python
dropoff_zones = zones.rename(columns={
    "LocationID": "DOLocationID",
    "Borough"   : "dropoff_borough",
    "Zone"      : "dropoff_zone"
})
```

```python
trips_clean = pd.merge(
    trips_clean,
    dropoff_zones[["DOLocationID", "dropoff_borough", "dropoff_zone"]],
    on="DOLocationID",
    how="left"
)
```

```python
unmatched_dropoff = trips_clean["dropoff_borough"].isnull().sum()
final_rows        = len(trips_clean)
```

```python
print(f"Unmatched dropoff zone rows: {unmatched_dropoff:,}")
print(f"Final rows after both joins: {final_rows:,}")
print(trips_clean[["DOLocationID", "dropoff_borough", "dropoff_zone"]].head())
```

```python
print("=" * 55)
print("          JOIN QUALITY REPORT")
print("=" * 55)
print(f"  Raw rows (pre-clean)          : {raw_rows:>10,}")
print(f"  Cleaned rows (valid trips)    : {cleaned_rows:>10,}")
print(f"  Unmatched pickup zones        : {unmatched_pickup:>10,}")
print(f"  Unmatched dropoff zones       : {unmatched_dropoff:>10,}")
print(f"  Final rows (after all joins)  : {final_rows:>10,}")
print("=" * 55)
```

```python
zone_summary = trips_clean.groupby(["pickup_borough", "pickup_zone"]).agg(
    trip_count    = ("total_amount", "count"),
    total_revenue = ("total_amount", "sum"),
    avg_revenue   = ("total_amount", "mean"),
    avg_distance  = ("trip_distance", "mean"),
    min_distance  = ("trip_distance", "min"),
    max_distance  = ("trip_distance", "max")
).reset_index()
```

```python
zone_summary = zone_summary.sort_values("total_revenue", ascending=False)
```

```python
print(f"Zone summary shape: {zone_summary.shape}")
print("\nTop 10 Pickup Zones by Revenue:")
display(zone_summary.head(10))
```

```python
payment_map = {1: "Credit Card", 2: "Cash", 3: "No Charge", 4: "Dispute", 5: "Unknown", 6: "Voided"}
trips_clean["payment_label"] = trips_clean["payment_type"].map(payment_map).fillna("Other")
```

```python
payment_summary = trips_clean.groupby(["pickup_month", "payment_label"]).agg(
    trip_count    = ("fare_amount", "count"),
    total_revenue = ("fare_amount", "sum"),
    avg_fare      = ("fare_amount", "mean"),
    min_fare      = ("fare_amount", "min"),
    max_fare      = ("fare_amount", "max")
).reset_index()
```

```python
print("Payment × Month Summary:")
display(payment_summary)
```

```python
# Pivot table — index=pickup_hour, columns=pickup_borough, values=total_amount
pivot = pd.pivot_table(
    trips_clean,
    values   = "total_amount",
    index    = "pickup_hour",
    columns  = "pickup_borough",
    aggfunc  = "sum",
    fill_value = 0
)
```

```python
# Clean up column name artifacts from pivot
pivot.columns.name = None
pivot = pivot.reset_index()
```

```python
print(f"Pivot table shape: {pivot.shape}")
print("\nBorough × Hour Revenue Pivot (first 5 rows):")
display(pivot.head())
```

```python
# Melt the wide pivot into long dashboard-ready format
long_report = pivot.melt(
    id_vars    = "pickup_hour",
    var_name   = "pickup_borough",
    value_name = "revenue"
)
```

```python
long_report = long_report.sort_values(["pickup_hour", "pickup_borough"]).reset_index(drop=True)
```

```python
print(f"Long report shape: {long_report.shape}")
print("\nLong format (first 10 rows):")
display(long_report.head(10))
```

```python
# Export all required output files
zone_summary.to_csv("pickup_zone_summary.csv", index=False)
pivot.to_csv("borough_hour_pivot.csv", index=False)
long_report.to_csv("borough_hour_long.csv", index=False)
payment_summary.to_csv("payment_month_summary.csv", index=False)
```

```python
print("\n" + "=" * 65)
print("BUSINESS QUESTION 4: Top 10 Pickup → Dropoff Routes")
print("=" * 65)
route_summary = (
    trips_clean
    .groupby(["pickup_zone", "dropoff_zone"])
    .size()
    .reset_index(name="trip_count")
    .sort_values("trip_count", ascending=False)
    .head(10)
)
display(route_summary)
```

```python
print("\n" + "=" * 65)
print("BUSINESS QUESTION 5: Revenue Change Jan → Feb")
print("=" * 65)
monthly_rev = trips_clean.groupby("pickup_month")["total_amount"].sum()
print(monthly_rev)
if len(monthly_rev) >= 2:
    jan_rev = monthly_rev[1]
    feb_rev = monthly_rev[2]
    change  = (feb_rev - jan_rev) / jan_rev * 100
    direction = "INCREASED ↑" if change > 0 else "DECREASED ↓"
    print(f"\n  January  Revenue : ${jan_rev:,.2f}")
    print(f"  February Revenue : ${feb_rev:,.2f}")
    print(f"  Change           : {change:+.2f}%  → Revenue {direction}")
```
