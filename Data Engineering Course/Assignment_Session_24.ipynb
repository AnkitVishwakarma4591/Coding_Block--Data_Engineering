```python
import requests
import pandas as pd
```

```python
orders = requests.get("https://fake-store-api.mock.beeceptor.com/api/orders").json()
users = requests.get("https://fake-store-api.mock.beeceptor.com/api/users").json()
products = requests.get("https://fake-store-api.mock.beeceptor.com/api/products").json()
```

## Orders Data

```python
# Convert to dataframe
df = pd.DataFrame(orders)

# Split list into rows
df = df.explode('items')

# Convert dictionary into columns
items_df = pd.json_normalize(df['items'])

# Combine both
orders_data = pd.concat(
    [df.drop(columns=['items']).reset_index(drop=True),
     items_df.reset_index(drop=True)],
    axis=1
)

print(orders_data.head())
```

## Users Data

```python
# Convert to dataframe
users_data = pd.DataFrame(users)
```

```python
users_data
```

## Products Data

```python
# Convert to dataframe
df = pd.DataFrame(products)

# Split list into rows
df = df.explode('reviews')

# Convert dictionary into columns
items_df = pd.json_normalize(df['reviews'])

# Combine both
product_data = pd.concat(
    [df.drop(columns=['reviews']).reset_index(drop=True),
     items_df.reset_index(drop=True)],
    axis=1
)
print(product_data.head())
```

```python
orders_data.to_csv('orders.csv', index=False)
users_data.to_csv('users.csv', index=False)
product_data.to_csv('product.csv', index=False)
```

# Joins performing on order and user table

### inner

```python
order_user_inner = pd.merge(
    order,
    product,
    on="user_id",
    how="inner"
)

print(order_user_inner)
```

### left

```python
order_user_left = pd.merge(
    order,
    product,
    on="user_id",
    how="left"
)

print(order_user_left)
```

### right

```python
order_user_right = pd.merge(
    order,
    product,
    on="user_id",
    how="right"
)

print(order_user_right)
```

### outer

```python
order_user_outer = pd.merge(
    order,
    product,
    on="user_id",
    how="outer"
)

print(order_user_outer)
```

### Join Indicator

```python
order_user_outer = pd.merge(
    order,
    product,
    on="user_id",
    how="outer",
    indicator=True
)

print(order_user_outer)
```
